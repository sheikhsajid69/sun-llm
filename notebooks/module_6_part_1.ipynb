```python
# Dataset
# "I am learning the Natural Language Processing and running the experiments efficiently"
import nltk
nltk.download('punkt') #used for the tokenization
nltk.download('stopwords') #providing the common stopword list
nltk.download('wordnet') # requires the lemmatization
```

```python
# Tokenization
from nltk.tokenize import word_tokenize
import nltk
nltk.download('punkt_tab')

text = "I am learning Natural Language Processing and running experiments efficiently."
tokens = word_tokenize(text)

print(tokens)
```

```python
#Stopword Removal
from nltk.corpus import stopwords

stop_words = set(stopwords.words('english'))
filtered_tokens = [word for word in tokens if word.lower() not in stop_words]
print(filtered_tokens)
```

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_tokens]

print(stemmed_words)
```

```python
# Lemmatizations
from nltk.stem import WordNetLemmatizer
lemmatizer = WordNetLemmatizer()
lemmatized_words = [
    lemmatizer.lemmatize(word, pos='v')
    for word in filtered_tokens
]
print(lemmatized_words)
```
