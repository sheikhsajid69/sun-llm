```python
# Vectorization Techniques : Bag of Words(BoW) , TF-IDF , Word Embedding (Word2Vec , GloVe)
documents = [
    "I love Artificial Intelligence",
    "Artificial Intelligence is amazing",
    "I love deep learning"
]
```

```python
from sklearn.feature_extraction.text import CountVectorizer
vectorizer = CountVectorizer()
bow_matrix = vectorizer.fit_transform(documents)
print("Vocabulary:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", bow_matrix.toarray())
```

```python
# TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer()
tfidf_matrix = tfidf.fit_transform(documents)

print("Vocabulary:", tfidf.get_feature_names_out())
print("TF-IDF Matrix:\n", tfidf_matrix.toarray())

# TF-IDF = (Term Frequency) × log(Total Docs / Docs containing word)
# Text Classification
# Spam Detection
# Search Engines
# Sentiment Analysis
```

```python
!pip install gensim
# Word2Vec
from gensim.models import Word2Vec
sentences = [
    "I love Artificial Intelligence",
    "Artificial Intelligence is amazing",
    "I love deep learning"
]

# Tokenize sentences
tokenized_sentences = [s.lower().split() for s in sentences]

model = Word2Vec(tokenized_sentences, vector_size=50, window=3, min_count=1)
vector = model.wv['learning']
print('vector size : ' , len(vector))
```
